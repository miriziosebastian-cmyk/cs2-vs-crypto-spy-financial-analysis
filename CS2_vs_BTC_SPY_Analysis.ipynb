```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Estilo visual limpio
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# 1. Cargar datos
file_path = 'Copia de case study.xlsx'
df_comp = pd.read_excel(file_path, sheet_name='asset_comparison')
df_metrics = pd.read_excel(file_path, sheet_name='metrics')

# Convertir la columna de mes a tipo fecha (datetime) para evitar warnings en el eje X
df_comp['date'] = pd.to_datetime(df_comp['month'])

print("=== MÉTRICAS RESUMEN ===")
print(df_metrics.to_string(index=False))

# 2. Generar Gráfico
plt.figure(figsize=(12, 6), dpi=300)

plt.plot(df_comp['date'], df_comp['gamma2_index_rare'], label='CS2 Gamma 2 Case', color='#E65100', linewidth=2.5)
plt.plot(df_comp['date'], df_comp['btc_index_rare'], label='Bitcoin (BTC)', color='#F57C00', linewidth=2, linestyle='--')
plt.plot(df_comp['date'], df_comp['spy_index_rare'], label='S&P 500 (SPY)', color='#1976D2', linewidth=2, linestyle=':')

plt.title('Asset Performance Comparison (Rebased June 2021 = 100)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Date', fontsize=11, labelpad=10)
plt.ylabel('Normalized Value (Base 100)', fontsize=11)

plt.legend(frameon=True, facecolor='white', edgecolor='none', fontsize=10)
plt.tight_layout()

# Guardar imagen en alta resolución para el README de GitHub
plt.savefig('asset_performance_comparison.png', dpi=300)
plt.show()
```
